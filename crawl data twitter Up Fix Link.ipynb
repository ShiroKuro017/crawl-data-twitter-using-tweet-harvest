Crawl data twitter menggunakan Tweet-Harvest
===


`filename = 'FERRY_IRWANDI.csv'`

filename merupakan layout penamaan [filename].csv

`search_keyword = 'ferry irwandi since:2025-01-01 until:2025-05-01 lang:id'`

merupakan keyword pencarian di Twitter / X dengan layout :
[Topik yang dicari] [batas tanggal awal data yang dicari] [batas tanggal akhir data yang dicari] [bahasa dari topik yang sedang di cari]

`limit = 100`

merupakan jumlah limit pencarian yang akan di ambil (kadang bisa lebih beberapa dari limit) 

`!npx -y tweet-harvest@2.6.1 -o "{filename}" -s "{search_keyword}" --tab "LATEST" -l {limit} --token {twitter_auth_token}`

script untuk menjalankan sistem tweet-harvest yang berbasis Nodejs


````python
# Crawl Data

filename = "TUNTUTAN_RAKYAT.csv"
search_keyword = "17+8 TUNTUTAN RAKYAT since:2025-09-01 until:2025-09-05 lang:id"
limit = 1500

!npx -y tweet-harvest@2.6.1 -o "{filename}" -s "{search_keyword}" --tab "LATEST" -l {limit} --token {twitter_auth_token}
````

Menampilkan data .CSV menjadi DataFrame
===

````python
import pandas as pd

# jelaskan lokasi file ke CSV file
file_path = f"tweets-data/{filename}"

# mambaca CSV file ke pandas DataFrame
df = pd.read_csv(file_path, delimiter=",")

# menampilkan DataFrame
display(df)
````

untuk membenarkan link url yang berubah karena Tweet-Harvest gagal mengambil Username maka perlu kita ubah menggunakan URL dasar Twitter/X seperti 
```
https://x.com/x/status/
```
kemudian di ikuti dengan data id_str seperti
```  
https://x.com/x/status/ + df['id_str']
```

## fix_link


````python
import numpy as np

# Membaca file CSV
df = pd.read_csv(file_path, delimiter=",")

# Memperbaiki URL yang mengandung 'undefined'
df['tweet_url'] = np.where(
    df['tweet_url'].str.contains('undefined', na=False),
    'https://x.com/anyuser/status/' + df['id_str'].astype(str),
    df['tweet_url']
)

# Menghapus kolom username dan location
df = df.drop(columns=['location', 'username']) # Hapus kolom yang tidak diperlukan

# Menyimpan hasil ke file CSV baru
df.to_csv('tweets-data/TUNTUTAN_RAKYAT_fix.csv', index=False)
display(df)
````
